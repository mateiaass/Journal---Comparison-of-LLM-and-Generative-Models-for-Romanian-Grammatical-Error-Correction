```python
import json

file_path = "sentences.jsonl"

results = []

import re

def extrage_output(text):
    pattern = r"<\|end_header_id\|>\s*(.*?)\s*<\|eot_id\|>"
    matches = re.findall(pattern, text, re.DOTALL)

    if matches:
        return matches[-1].strip()
    return None

with open(file_path, "r", encoding="utf-8") as f:
    for line in f:
        data = json.loads(line)
        results.append((data[0], data[1], extrage_output(data[2])))
```

```python
import json

file_name = f'./rollama3/sentences.jsonl'
with open(file_name, 'w', encoding='utf-8') as f:
    for pair in results:
        json.dump(pair, f, ensure_ascii=False)
        f.write('\n')
```

```python
def read_jsonl_files(directory):
    sentence_pairs = []
    for filename in os.listdir(directory):
        if filename.endswith('.jsonl'):
            file_path = os.path.join(directory, filename)
            with open(file_path, 'r', encoding='utf-8') as f:
                for line in f:
                    data = json.loads(line.strip())

                    if isinstance(data, list) and len(data) == 3:
                        correct_sentence, input_sentence, predicted_sentence = data
                        sentence_pairs.append((correct_sentence, input_sentence, predicted_sentence))
    return sentence_pairs
```

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')

def tokenize_sentence(sentence):
    tokens = word_tokenize(sentence)
    return ' '.join(tokens)
```

```python
def write_sentences_to_file(sentences, file_path):
    with open(file_path, 'w', encoding='utf-8') as f:
        for sentence in sentences:
            tokenized_sentence = tokenize_sentence(sentence)
            f.write(tokenized_sentence + '\n')
```

```python
import os

directory = './rollama3'
sentence_pairs = read_jsonl_files(directory)

predicted_sentences = [pair[2] for pair in sentence_pairs]
wrong_sentences = [pair[1] for pair in sentence_pairs]
original_sentences = [pair[0] for pair in sentence_pairs]

print(len(wrong_sentences))
print(len(original_sentences))

write_sentences_to_file(wrong_sentences, './rollama3/wrong_file.txt')
write_sentences_to_file(original_sentences, './rollama3/original_file.txt')
write_sentences_to_file(predicted_sentences, './rollama3/predicted_file.txt')
```
