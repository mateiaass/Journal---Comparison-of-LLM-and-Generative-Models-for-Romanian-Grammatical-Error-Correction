```python
from datasets import load_dataset

dataset = load_dataset("upb-nlp/gec-ro-comments")
dataset
```

```python
dataset = dataset["train"]
```

```python
paired_dataset = []

for i in range(0, len(dataset) - 1, 2):
    wrong = dataset[i]["text"]
    right = dataset[i + 1]["text"]
    paired_dataset.append({
        "wrong": wrong,
        "right": right
    })

for pair in paired_dataset:
    print(pair)
```

```python
from datasets import Dataset

hf_dataset = Dataset.from_list(paired_dataset)
hf_dataset
```
